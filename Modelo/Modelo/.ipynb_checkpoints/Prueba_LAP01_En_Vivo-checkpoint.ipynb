# LAP01 · prueba en vivo de las tres cámaras

Ejecuta las celdas en orden en **JupyterLab del entorno `jupyterev`**. Verás las tres cámaras, una debajo de otra, actualizándose mientras se detectan y siguen las personas.

Se exportó la **clase final efectiva de LAP01**, con sus pesos YOLO26m, ResNet18, memoria visual, Hungarian y parámetros originales. Una instancia de tracking por cámara comparte los modelos de inferencia.

**Alcance actual:** los rótulos `cam01 L1`, `cam02 L1`, etc. son IDs locales. LAP01 no contiene asociación global entre cámaras. `track_uid` combina sesión, cámara e ID local; `global_id` queda vacío. No implica que dos personas con el mismo número local sean la misma persona.


## 1. Preparar el entorno y las rutas

Usa el kernel **Python (jupyterev)**. Si abriste JupyterLab desde ese entorno, su kernel Python también puede aparecer como **Python 3 (ipykernel)**. La celda muestra el intérprete utilizado. No se instalan ni descargan modelos al ejecutar esta prueba.


```python
from pathlib import Path
import os
import sys

# Funciona abriendo JupyterLab desde esta carpeta o desde Entrenamiento.
candidates = [Path.cwd(), Path.cwd() / '02_multicamara' / 'Prueba_LAP01',
              Path.cwd() / 'Prueba_LAP01']
BASE = next((p.resolve() for p in candidates if (p / 'modelo' / 'config_lap01.json').is_file()), None)
if BASE is None:
    raise FileNotFoundError('Abre JupyterLab desde la carpeta Prueba_LAP01 o Entrenamiento.')
if str(BASE) not in sys.path:
    sys.path.insert(0, str(BASE))

# Configuración de Ultralytics dentro de la carpeta de prueba.
os.environ.setdefault('YOLO_CONFIG_DIR', str(BASE / '.runtime' / 'ultralytics'))
os.environ.setdefault('YOLO_AUTOINSTALL', 'false')

import torch
import pandas as pd
from IPython.display import display, Image
from aerovision import MotorLAP01, inspeccionar_videos, procesar_videos
from aerovision.notebook import VistaEnVivo

print('Python:', sys.executable)
print('Carpeta:', BASE)
print('GPU:', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU')
if Path(sys.prefix).name != 'jupyterev':
    raise RuntimeError('Selecciona el kernel de jupyterev antes de continuar.')
```

## 2. Configurar la prueba

Los tres videos tienen 1280×720, 15 FPS y 1892 frames (~126 segundos). `MAX_FRAMES = None` procesa el video completo; usa `450` para probar 30 segundos.

`TIEMPO_REAL = True` respeta el ritmo original cuando la máquina tiene capacidad. Si el cálculo tarda más, la vista avanza a la velocidad alcanzable **sin saltar frames**. El contador distingue FPS por cámara y FPS totales.

`VISTA_FPS` controla las actualizaciones de la imagen en JupyterLab, no la frecuencia del tracker. Los videos comienzan en el primer frame; tener la misma duración no demuestra sincronización física. `INICIOS` permite omitir un prefijo si necesitas alinearlos manualmente.


```python
VIDEOS = {
    cid: BASE.parent / 'videocamara' / f'{cid}_recortado_1280x720_15fps.mp4'
    for cid in ('cam01', 'cam02', 'cam03')
}
MAX_FRAMES = None
TIEMPO_REAL = True
VISTA_FPS = 10.0
BATCH_YOLO = True                # Un lote con un frame de cada cámara; mismo tracker.
DEVICE = 'auto'                  # cuda:0 si está disponible; CPU en caso contrario.
INICIOS = {'cam01': 0, 'cam02': 0, 'cam03': 0}  # Índices de inicio, base cero.

display(pd.DataFrame(inspeccionar_videos(VIDEOS)))
```

## 3. Cargar los modelos una vez

Las siguientes ejecuciones de la celda de procesamiento reutilizan estos pesos y reinician los estados de tracking. Conserva `BATCH_YOLO = False` si quieres usar la misma modalidad de inferencia individual del notebook original; con `True` se agrupan las cámaras para aprovechar la GPU. Se mantiene FP32 y la resolución de inferencia original.


```python
motor = MotorLAP01(BASE / 'modelo', device=DEVICE, batch=BATCH_YOLO)
print('Modelos listos:', motor.device)
print('Parámetros originales de tracking:')
display(pd.Series(motor.config['tracker']))
```

## 4. Ver el procesamiento en vivo

Al ejecutar esta celda, las tres vistas aparecen **debajo de ella, en vertical (cam01, cam02, cam03), y se actualizan durante el análisis**. El estado `[*]` del kernel es normal mientras procesa. El arranque calienta la GPU antes de comenzar a medir.

Para detener antes del final, usa **■ Interrupt Kernel** en JupyterLab. Se cierran los videos y los resultados parciales quedan en `resultado`. Para repetir, vuelve a ejecutar esta celda. **No se guarda ningún archivo**: el resultado solo se muestra en el notebook.


```python
vista = VistaEnVivo()
resultado = procesar_videos(
    motor, VIDEOS,
    max_frames=MAX_FRAMES,
    tiempo_real=TIEMPO_REAL,
    preview=vista,
    preview_fps=VISTA_FPS,
    start_frames=INICIOS,
)
vista.terminar(resultado)
```

## 5. Consultar los resultados

`resultado.tabla()` contiene cámara, frame original, tiempo relativo, ID local, caja y confianza. Todo vive en memoria: al reiniciar el kernel se pierde.

Los conteos son **tracks confirmados**, no una medida validada de personas únicas: una identidad puede fragmentarse. Medir precisión requiere anotaciones de referencia.


```python
print('Sesión:', resultado.resumen['session_id'], '·', resultado.resumen['status'])
print(f"Capacidad medida: {resultado.resumen['fps_per_camera']:.2f} FPS/cámara "
      f"({resultado.resumen['fps_total']:.2f} FPS totales)")
display(pd.DataFrame(resultado.resumen['cameras']).T)
tracks = resultado.tabla()
display(tracks.head(20))
if not tracks.empty:
    display(tracks.groupby('camera_id').agg(ids_locales=('local_id', 'nunique'),
                                           observaciones=('local_id', 'size')))
```
